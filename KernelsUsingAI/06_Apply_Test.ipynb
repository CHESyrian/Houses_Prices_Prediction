<div class="nb-title">
  <h1>06 — Apply Models on Test Data</h1>
  <p class="subtitle">Generate final predictions and submission file</p>
  <p class="meta">Ensemble scoring · expm1 transform · Sanity checks · Delivery</p>
</div>


```python
import pandas as pd
import numpy as np
from pathlib import Path
import joblib
import warnings
warnings.filterwarnings('ignore')

DATA = Path('../Datasets/processed')
MODEL_DIR = Path('../Reports/models')
SUB_DIR = Path('../Reports')

X_test = pd.read_csv(DATA / 'X_test_selected.csv')
test_id = pd.read_csv(DATA / 'test_id.csv').squeeze()

print('Test shape:', X_test.shape)
print('Test IDs:', len(test_id))

hgb = joblib.load(MODEL_DIR / 'best_hgb.joblib')
ridge = joblib.load(MODEL_DIR / 'best_ridge.joblib')
rf = joblib.load(MODEL_DIR / 'rf.joblib')
gbm = joblib.load(MODEL_DIR / 'gbm.joblib')

pred_hgb = hgb.predict(X_test)
pred_ridge = ridge.predict(X_test)
pred_rf = rf.predict(X_test)
pred_gbm = gbm.predict(X_test)

pred_log = (pred_hgb + pred_ridge + pred_rf + pred_gbm) / 4
pred_price = np.expm1(pred_log)

print('Predicted price stats:')
print(pd.Series(pred_price).describe())

submission = pd.DataFrame({'Id': test_id, 'SalePrice': pred_price})
submission.to_csv(SUB_DIR / 'submission.csv', index=False)
print('\nSubmission saved to Reports/submission.csv')
print(submission.head(10))

pd.DataFrame({
    'Id': test_id,
    'hgb': np.expm1(pred_hgb),
    'ridge': np.expm1(pred_ridge),
    'rf': np.expm1(pred_rf),
    'gbm': np.expm1(pred_gbm),
    'ensemble': pred_price
}).to_csv(SUB_DIR / 'test_predictions_detail.csv', index=False)
print('Detailed predictions saved')
```

<div class="nb-section">
  <h2>Sanity Checks</h2>
</div>


```python
y_train = pd.read_csv(DATA / 'y.csv').squeeze()
print('Train SalePrice mean/median:', y_train.mean(), y_train.median())
print('Pred  SalePrice mean/median:', pred_price.mean(), np.median(pred_price))

import matplotlib.pyplot as plt
import seaborn as sns
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(y_train, kde=True, ax=axes[0], color='steelblue', label='Train')
sns.histplot(pred_price, kde=True, ax=axes[0], color='coral', label='Test Pred', alpha=0.6)
axes[0].legend()
axes[0].set_title('SalePrice Distribution: Train vs Pred')
axes[1].boxplot([y_train, pred_price], labels=['Train', 'Pred'])
axes[1].set_title('Boxplot comparison')
plt.tight_layout()
plt.savefig('../Reports/model_figures/03_pred_distribution.png', bbox_inches='tight')
plt.show()
```
