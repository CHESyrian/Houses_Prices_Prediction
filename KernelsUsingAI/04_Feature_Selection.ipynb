<div class="nb-title">
  <h1>04 — Feature Selection</h1>
  <p class="subtitle">Reducing dimensionality while preserving predictive power</p>
  <p class="meta">Variance filter · Correlation filter · Mutual information · RFE · Validation</p>
</div>


```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.feature_selection import VarianceThreshold, mutual_info_regression, RFE
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

DATA = Path('../Datasets/processed')
OUT = Path('../Datasets/processed')
FIG = Path('../Reports/selection_figures')
FIG.mkdir(parents=True, exist_ok=True)

X = pd.read_csv(DATA / 'X_train.csv')
y_log = pd.read_csv(DATA / 'y_log.csv').squeeze()
X_test = pd.read_csv(DATA / 'X_test.csv')

print('Starting features:', X.shape[1])
```

<div class="nb-box">
<strong>Methods used</strong><br/>1. Variance threshold (remove near-zero variance)<br/>2. Correlation filter (remove highly collinear features)<br/>3. Mutual information / univariate ranking<br/>4. Recursive Feature Elimination with Random Forest (RFE)<br/>5. Final selected feature set saved for modeling
</div>


```python
# Remove features with variance < 0.01 (after scaling would be different, but here raw)
selector = VarianceThreshold(threshold=0.01)
selector.fit(X)
mask_var = selector.get_support()
X_var = X.loc[:, mask_var]
print(f'After variance filter: {X_var.shape[1]} (removed {X.shape[1]-X_var.shape[1]})')
```

<div class="nb-section">
  <h2>1. Variance Threshold</h2>
</div>


```python
# Remove one of each pair with |corr| > 0.95
corr_matrix = X_var.corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop = [col for col in upper.columns if any(upper[col] > 0.95)]
print(f'Highly correlated features to drop ({len(to_drop)}):', to_drop[:15])
X_corr = X_var.drop(columns=to_drop)
print(f'After correlation filter: {X_corr.shape[1]}')
```

<div class="nb-section">
  <h2>2. High Correlation Filter</h2>
</div>


```python
mi = mutual_info_regression(X_corr, y_log, random_state=42)
mi_series = pd.Series(mi, index=X_corr.columns).sort_values(ascending=False)
print(mi_series.head(20))

fig, ax = plt.subplots(figsize=(9, 7))
mi_series.head(25).plot.barh(ax=ax, color='darkorange')
ax.set_title('Mutual Information with log(SalePrice) – Top 25')
ax.invert_yaxis()
plt.tight_layout()
plt.savefig(FIG / '01_mutual_info.png', bbox_inches='tight')
plt.show()
```

<div class="nb-section">
  <h2>3. Mutual Information Ranking</h2>
</div>


```python
# Select top k features via RFE
k = 80  # target number of features
rf = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42, n_jobs=-1)
rfe = RFE(estimator=rf, n_features_to_select=k, step=5)
rfe.fit(X_corr, y_log)

selected = X_corr.columns[rfe.support_].tolist()
print(f'RFE selected {len(selected)} features')
print(selected[:30])

# Ranking
ranking = pd.Series(rfe.ranking_, index=X_corr.columns).sort_values()
print('\nTop ranked (1 = selected):')
print(ranking.head(15))
```

<div class="nb-section">
  <h2>4. RFE with Random Forest</h2>
</div>


```python
X_sel = X_corr[selected]
X_test_sel = X_test.reindex(columns=selected, fill_value=0)

rf_final = RandomForestRegressor(n_estimators=200, max_depth=12, random_state=42, n_jobs=-1)
scores = cross_val_score(rf_final, X_sel, y_log, cv=5,
                         scoring='neg_root_mean_squared_error', n_jobs=-1)
print(f'Selected set 5-fold CV RMSE (log): {-scores.mean():.5f} ± {scores.std():.5f}')

# Compare with full set (after corr filter)
scores_full = cross_val_score(rf_final, X_corr, y_log, cv=5,
                              scoring='neg_root_mean_squared_error', n_jobs=-1)
print(f'Full (post-corr) CV RMSE (log): {-scores_full.mean():.5f} ± {scores_full.std():.5f}')
```

<div class="nb-section">
  <h2>5. Validate selected set with CV</h2>
</div>


```python
X_sel.to_csv(OUT / 'X_train_selected.csv', index=False)
X_test_sel.to_csv(OUT / 'X_test_selected.csv', index=False)
pd.Series(selected).to_csv(OUT / 'selected_features.csv', index=False, header=['feature'])
print('Saved selected feature matrices and list')
print('Final feature count:', len(selected))
```
