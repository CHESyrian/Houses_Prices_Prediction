  <h1>01 — Exploratory Data Analysis</h1>

<div class="nb-section">
  <h2>1. Setup & Data Loading</h2>
</div>


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# Paths
DATA_DIR = Path('../Datasets')
OUT_DIR = Path('../Reports/eda_figures')
OUT_DIR.mkdir(parents=True, exist_ok=True)

pd.set_option('display.max_columns', 100)
pd.set_option('display.max_rows', 100)
sns.set_theme(style='whitegrid', palette='muted', font_scale=1.05)
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['figure.dpi'] = 100

train = pd.read_csv(DATA_DIR / 'train.csv')
test  = pd.read_csv(DATA_DIR / 'test.csv')

print(f'Train: {train.shape[0]:,} rows × {train.shape[1]} cols')
print(f'Test : {test.shape[0]:,} rows × {test.shape[1]} cols')
print(f'Target: SalePrice (present in train only)')
train.head(3)
```

<div class="nb-section">
  <h2>2. Schema & Data Types</h2>
</div>


```python
print('=== Data Types ===')
print(train.dtypes.value_counts())
print('\n=== Numeric columns ===')
num_cols = train.select_dtypes(include=[np.number]).columns.tolist()
print(len(num_cols), num_cols)
print('\n=== Categorical / Object columns ===')
cat_cols = train.select_dtypes(include=['object']).columns.tolist()
print(len(cat_cols), cat_cols)
```

<div class="nb-section">
  <h2>3. Missing Values Analysis</h2>
</div>


```python
def missing_report(df, name='df'):
    miss = df.isnull().sum()
    miss = miss[miss > 0].sort_values(ascending=False)
    pct = (miss / len(df) * 100).round(2)
    report = pd.DataFrame({'Missing': miss, 'Pct': pct})
    print(f'{name}: {len(miss)} columns with missing values')
    return report

print('--- TRAIN ---')
miss_train = missing_report(train, 'Train')
display(miss_train)

print('\n--- TEST ---')
miss_test = missing_report(test, 'Test')
display(miss_test.head(20))
```

```python
# Visualize missingness (top 20)
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
miss_train.head(20).Pct.plot.barh(ax=axes[0], color='steelblue')
axes[0].set_title('Train – Missing % (top 20)')
axes[0].set_xlabel('% Missing')
miss_test.head(20).Pct.plot.barh(ax=axes[1], color='coral')
axes[1].set_title('Test – Missing % (top 20)')
axes[1].set_xlabel('% Missing')
plt.tight_layout()
plt.savefig(OUT_DIR / '01_missingness.png', bbox_inches='tight')
plt.show()
```

<div class="nb-callout">
<strong>Key observations on missingness</strong><br/>• <code>PoolQC</code>, <code>MiscFeature</code>, <code>Alley</code>, <code>Fence</code> are mostly missing → often mean “None” (no pool / no alley / no fence).<br/>• Garage- and Basement-related fields share the same missing pattern (no garage / no basement).<br/>• <code>LotFrontage</code> (~18%) and <code>MasVnrType</code>/<code>MasVnrArea</code> need careful imputation.<br/>• Only 1 missing value in <code>Electrical</code> (train).
</div>


<div class="nb-section">
  <h2>4. Target Variable — SalePrice</h2>
</div>


```python
print(train['SalePrice'].describe())
print(f'\nSkewness: {train["SalePrice"].skew():.3f}')
print(f'Kurtosis: {train["SalePrice"].kurtosis():.3f}')

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(train['SalePrice'], kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('SalePrice Distribution')
sns.histplot(np.log1p(train['SalePrice']), kde=True, ax=axes[1], color='seagreen')
axes[1].set_title('log1p(SalePrice)')
sns.boxplot(x=train['SalePrice'], ax=axes[2], color='steelblue')
axes[2].set_title('SalePrice Boxplot')
plt.tight_layout()
plt.savefig(OUT_DIR / '02_saleprice_dist.png', bbox_inches='tight')
plt.show()
```

<div class="nb-callout">
<strong>Insight:</strong> SalePrice is right-skewed (skew ≈ 1.88). Log-transform will be used for modeling to stabilize variance and improve residual normality.
</div>


<div class="nb-section">
  <h2>5. Numeric Feature Distributions & Correlations</h2>
</div>


```python
# Key numeric features
key_num = ['OverallQual', 'GrLivArea', 'GarageCars', 'GarageArea', 'TotalBsmtSF',
           '1stFlrSF', 'FullBath', 'TotRmsAbvGrd', 'YearBuilt', 'YearRemodAdd',
           'LotArea', 'LotFrontage', 'MasVnrArea', 'Fireplaces']

# Correlation with target
corr_target = train[key_num + ['SalePrice']].corr()['SalePrice'].sort_values(ascending=False)
print('Correlation with SalePrice:')
print(corr_target)

fig, ax = plt.subplots(figsize=(8, 6))
corr_target.drop('SalePrice').plot.barh(ax=ax, color='steelblue')
ax.set_title('Pearson Correlation with SalePrice')
ax.set_xlabel('Correlation')
plt.tight_layout()
plt.savefig(OUT_DIR / '03_corr_target.png', bbox_inches='tight')
plt.show()
```

```python
# Full correlation heatmap (numeric only, top correlated)
num_df = train.select_dtypes(include=[np.number]).drop(columns=['Id'], errors='ignore')
corr = num_df.corr()
# Focus on features with |corr| > 0.3 with SalePrice
top_corr_cols = corr['SalePrice'].abs().sort_values(ascending=False).head(16).index.tolist()
fig, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(corr.loc[top_corr_cols, top_corr_cols], annot=True, fmt='.2f', cmap='RdBu_r',
            center=0, square=True, ax=ax, cbar_kws={'shrink': 0.8})
ax.set_title('Correlation Heatmap – Top Features vs SalePrice')
plt.tight_layout()
plt.savefig(OUT_DIR / '04_corr_heatmap.png', bbox_inches='tight')
plt.show()
```

<div class="nb-section">
  <h2>6. Key Relationships (Scatter / Box plots)</h2>
</div>


```python
# GrLivArea vs SalePrice (with potential outliers)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.scatterplot(data=train, x='GrLivArea', y='SalePrice', alpha=0.5, ax=axes[0])
axes[0].set_title('GrLivArea vs SalePrice')
# Highlight large houses with low price
outliers = train[(train['GrLivArea'] > 4000) & (train['SalePrice'] < 300000)]
axes[0].scatter(outliers['GrLivArea'], outliers['SalePrice'], color='red', s=80, label='Potential outliers')
axes[0].legend()

sns.boxplot(data=train, x='OverallQual', y='SalePrice', ax=axes[1], palette='Blues')
axes[1].set_title('OverallQual vs SalePrice')
plt.tight_layout()
plt.savefig(OUT_DIR / '05_key_relationships.png', bbox_inches='tight')
plt.show()
print(f'Potential outliers (GrLivArea > 4000 & SalePrice < 300k): {len(outliers)} rows')
display(outliers[['Id', 'GrLivArea', 'SalePrice', 'OverallQual']])
```

```python
# Neighborhood effect
order = train.groupby('Neighborhood')['SalePrice'].median().sort_values(ascending=False).index
fig, ax = plt.subplots(figsize=(14, 5))
sns.boxplot(data=train, x='Neighborhood', y='SalePrice', order=order, ax=ax, palette='viridis')
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
ax.set_title('SalePrice by Neighborhood (ordered by median)')
plt.tight_layout()
plt.savefig(OUT_DIR / '06_neighborhood.png', bbox_inches='tight')
plt.show()
```

<div class="nb-section">
  <h2>7. Categorical Feature Cardinality & Target Means</h2>
</div>


```python
# Cardinality
card = train[cat_cols].nunique().sort_values(ascending=False)
print('Categorical cardinality:')
print(card)

# Mean SalePrice for a few key cats
for col in ['ExterQual', 'KitchenQual', 'BsmtQual', 'GarageFinish', 'MSZoning']:
    print(f'\n--- {col} ---')
    print(train.groupby(col)['SalePrice'].agg(['count', 'mean', 'median']).sort_values('mean', ascending=False))
```

<div class="nb-section">
  <h2>8. Temporal Patterns (YearBuilt, YrSold, MoSold)</h2>
</div>


```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.scatterplot(data=train, x='YearBuilt', y='SalePrice', alpha=0.4, ax=axes[0])
axes[0].set_title('YearBuilt vs SalePrice')
train.groupby('YrSold')['SalePrice'].median().plot(ax=axes[1], marker='o')
axes[1].set_title('Median SalePrice by YrSold')
axes[1].set_ylabel('Median SalePrice')
train.groupby('MoSold')['SalePrice'].median().plot(ax=axes[2], marker='o', color='coral')
axes[2].set_title('Median SalePrice by MoSold')
axes[2].set_xlabel('Month')
plt.tight_layout()
plt.savefig(OUT_DIR / '07_temporal.png', bbox_inches='tight')
plt.show()
```

<div class="nb-section">
  <h2>9. Duplicate & Integrity Checks</h2>
</div>


```python
print('Duplicate rows (all columns):', train.duplicated().sum())
print('Duplicate Id:', train['Id'].duplicated().sum())
print('Id range train:', train['Id'].min(), '–', train['Id'].max())
print('Id range test :', test['Id'].min(), '–', test['Id'].max())
print('Overlap Ids:', len(set(train['Id']) & set(test['Id'])))

# Zero / negative checks for area columns
area_cols = ['LotArea', 'GrLivArea', 'TotalBsmtSF', '1stFlrSF', '2ndFlrSF', 'GarageArea']
for c in area_cols:
    n_zero = (train[c] == 0).sum()
    n_neg = (train[c] < 0).sum()
    print(f'{c}: zeros={n_zero}, negatives={n_neg}')
```
