<div class="nb-title">
  <h1>05 — Model Training</h1>
  <p class="subtitle">Train multiple regressors on the selected feature set</p>
  <p class="meta">Ridge · Lasso · ElasticNet · RandomForest · GradientBoosting · HistGradientBoosting · Ensemble</p>
</div>


```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import cross_val_score, KFold, GridSearchCV
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, HistGradientBoostingRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import joblib
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

DATA = Path('../Datasets/processed')
MODEL_DIR = Path('../Reports/models')
MODEL_DIR.mkdir(parents=True, exist_ok=True)
FIG = Path('../Reports/model_figures')
FIG.mkdir(parents=True, exist_ok=True)

X = pd.read_csv(DATA / 'X_train_selected.csv')
y_log = pd.read_csv(DATA / 'y_log.csv').squeeze()
y = pd.read_csv(DATA / 'y.csv').squeeze()

print(X.shape, y_log.shape)
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

<div class="nb-box">
<strong>Evaluation:</strong> 5-fold CV RMSE on <code>log(SalePrice)</code>. Final models saved with joblib. Ensemble = simple average of top models.
</div>


```python
# Linear models need scaling
ridge = Pipeline([
    ('scaler', StandardScaler()),
    ('model', Ridge(alpha=10.0, random_state=42))
])
lasso = Pipeline([
    ('scaler', StandardScaler()),
    ('model', Lasso(alpha=0.001, max_iter=5000, random_state=42))
])
enet = Pipeline([
    ('scaler', StandardScaler()),
    ('model', ElasticNet(alpha=0.001, l1_ratio=0.5, max_iter=5000, random_state=42))
])

rf = RandomForestRegressor(n_estimators=300, max_depth=15, min_samples_leaf=2,
                           random_state=42, n_jobs=-1)
gbm = GradientBoostingRegressor(n_estimators=300, learning_rate=0.05, max_depth=4,
                                min_samples_leaf=3, random_state=42)
hgb = HistGradientBoostingRegressor(max_iter=300, learning_rate=0.05, max_depth=6,
                                    random_state=42)

models = {
    'Ridge': ridge,
    'Lasso': lasso,
    'ElasticNet': enet,
    'RandomForest': rf,
    'GradientBoosting': gbm,
    'HistGradientBoosting': hgb,
}
```

<div class="nb-section">
  <h2>1. Define Models & Pipelines</h2>
</div>


```python
results = {}
for name, model in models.items():
    scores = cross_val_score(model, X, y_log, cv=kf,
                             scoring='neg_root_mean_squared_error', n_jobs=-1)
    rmse = -scores.mean()
    std = scores.std()
    results[name] = {'CV_RMSE_log': rmse, 'std': std}
    print(f'{name:22s}  RMSE(log) = {rmse:.5f} ± {std:.5f}')

res_df = pd.DataFrame(results).T.sort_values('CV_RMSE_log')
print('\nRanked by CV RMSE:')
display(res_df)

fig, ax = plt.subplots(figsize=(9, 5))
res_df['CV_RMSE_log'].plot.barh(ax=ax, color='steelblue', xerr=res_df['std'])
ax.set_xlabel('5-fold CV RMSE (log scale)')
ax.set_title('Model Comparison – Lower is better')
plt.tight_layout()
plt.savefig(FIG / '01_model_comparison.png', bbox_inches='tight')
plt.show()
```

<div class="nb-section">
  <h2>2. Cross-Validation Comparison</h2>
</div>


```python
# Tune HistGradientBoosting (reduced grid for speed)
param_grid = {
    'learning_rate': [0.05, 0.08],
    'max_depth': [5, 7],
    'max_iter': [300],
    'min_samples_leaf': [8],
}
hgb_base = HistGradientBoostingRegressor(random_state=42)
gs = GridSearchCV(hgb_base, param_grid, cv=kf,
                  scoring='neg_root_mean_squared_error', n_jobs=-1, verbose=1)
gs.fit(X, y_log)
print('Best params:', gs.best_params_)
print('Best CV RMSE:', -gs.best_score_)

best_hgb = gs.best_estimator_
```

```python
# Tune Ridge alpha
ridge_grid = Pipeline([
    ('scaler', StandardScaler()),
    ('model', Ridge(random_state=42))
])
gs_ridge = GridSearchCV(ridge_grid, {'model__alpha': [0.1, 1, 5, 10, 20, 50, 100]},
                        cv=kf, scoring='neg_root_mean_squared_error', n_jobs=-1)
gs_ridge.fit(X, y_log)
print('Best Ridge alpha:', gs_ridge.best_params_)
print('Best Ridge CV RMSE:', -gs_ridge.best_score_)
best_ridge = gs_ridge.best_estimator_
```

<div class="nb-section">
  <h2>3. Hyperparameter Tuning for Best Models</h2>
</div>


```python
# Fit top models on full train
best_hgb.fit(X, y_log)
best_ridge.fit(X, y_log)
rf.fit(X, y_log)
gbm.fit(X, y_log)

# OOF predictions for stacking/averaging evaluation
from sklearn.model_selection import cross_val_predict
oof_hgb = cross_val_predict(best_hgb, X, y_log, cv=kf, n_jobs=-1)
oof_ridge = cross_val_predict(best_ridge, X, y_log, cv=kf, n_jobs=-1)
oof_rf = cross_val_predict(rf, X, y_log, cv=kf, n_jobs=-1)
oof_gbm = cross_val_predict(gbm, X, y_log, cv=kf, n_jobs=-1)

oof_avg = (oof_hgb + oof_ridge + oof_rf + oof_gbm) / 4
rmse_avg = np.sqrt(mean_squared_error(y_log, oof_avg))
print(f'Simple average OOF RMSE (log): {rmse_avg:.5f}')

# Also report original-scale metrics for the average
pred_price = np.expm1(oof_avg)
print(f'OOF MAE (original $): {mean_absolute_error(y, pred_price):,.0f}')
print(f'OOF RMSE (original $): {np.sqrt(mean_squared_error(y, pred_price)):,.0f}')
print(f'OOF R²: {r2_score(y, pred_price):.4f}')
```

<div class="nb-section">
  <h2>4. Ensemble (Simple Average of Top Models)</h2>
</div>


```python
residuals = y_log - oof_avg
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(residuals, kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('OOF Residuals (log)')
axes[1].scatter(oof_avg, residuals, alpha=0.3, s=12)
axes[1].axhline(0, color='red', ls='--')
axes[1].set_xlabel('Predicted log(SalePrice)')
axes[1].set_ylabel('Residual')
axes[1].set_title('Residuals vs Fitted')
plt.tight_layout()
plt.savefig(FIG / '02_residuals.png', bbox_inches='tight')
plt.show()
```

<div class="nb-section">
  <h2>5. Residual Diagnostics</h2>
</div>


```python
joblib.dump(best_hgb, MODEL_DIR / 'best_hgb.joblib')
joblib.dump(best_ridge, MODEL_DIR / 'best_ridge.joblib')
joblib.dump(rf, MODEL_DIR / 'rf.joblib')
joblib.dump(gbm, MODEL_DIR / 'gbm.joblib')

# Save OOF and metrics
pd.DataFrame({
    'y_log': y_log,
    'oof_hgb': oof_hgb,
    'oof_ridge': oof_ridge,
    'oof_rf': oof_rf,
    'oof_gbm': oof_gbm,
    'oof_avg': oof_avg
}).to_csv(MODEL_DIR / 'oof_predictions.csv', index=False)

res_df.to_csv(MODEL_DIR / 'cv_results.csv')
print('Models and results saved to', MODEL_DIR)
```
