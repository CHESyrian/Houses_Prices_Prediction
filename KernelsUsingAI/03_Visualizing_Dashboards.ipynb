<div class="nb-title">
  <h1>03 — Visualizing & Dashboards</h1>
  <p class="subtitle">Post Feature-Engineering Insights</p>
  <p class="meta">Explore engineered features · Target relationships · Segments · Importance preview</p>
</div>


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

DATA = Path('../Datasets/processed')
FIG = Path('../Reports/viz_figures')
FIG.mkdir(parents=True, exist_ok=True)

X_train = pd.read_csv(DATA / 'X_train.csv')
y = pd.read_csv(DATA / 'y.csv').squeeze()
y_log = pd.read_csv(DATA / 'y_log.csv').squeeze()

sns.set_theme(style='whitegrid', font_scale=1.05)
plt.rcParams['figure.dpi'] = 100

print(X_train.shape, y.shape)
```

<div class="nb-section">
  <h2>1. Target after log transform</h2>
</div>


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(y, kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('Original SalePrice')
sns.histplot(y_log, kde=True, ax=axes[1], color='seagreen')
axes[1].set_title('log1p(SalePrice)')
plt.tight_layout()
plt.savefig(FIG / '01_target_log.png', bbox_inches='tight')
plt.show()
print(f'Skew original: {y.skew():.3f} | Skew log: {y_log.skew():.3f}')
```

<div class="nb-section">
  <h2>2. Engineered Feature vs Target</h2>
</div>


```python
eng_feats = ['HouseAge', 'TotalSF', 'TotalBath', 'TotalPorchSF', 'OverallGrade', 'RemodAge']
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
for ax, col in zip(axes.ravel(), eng_feats):
    if col in X_train.columns:
        sns.scatterplot(x=X_train[col], y=y_log, alpha=0.35, ax=ax, s=15)
        ax.set_title(f'{col} vs log(SalePrice)')
plt.tight_layout()
plt.savefig(FIG / '02_engineered_vs_target.png', bbox_inches='tight')
plt.show()
```

<div class="nb-section">
  <h2>3. Binary Flags Impact</h2>
</div>


```python
flags = ['HasPool', 'HasGarage', 'HasBsmt', 'HasFireplace', 'Has2ndFloor', 'WasRemodeled']
fig, axes = plt.subplots(2, 3, figsize=(14, 8))
for ax, col in zip(axes.ravel(), flags):
    if col in X_train.columns:
        sns.boxplot(x=X_train[col], y=y, ax=ax, palette='Set2')
        ax.set_title(col)
plt.tight_layout()
plt.savefig(FIG / '03_binary_flags.png', bbox_inches='tight')
plt.show()
```

<div class="nb-section">
  <h2>4. Top Correlated Features (post-FE)</h2>
</div>


```python
# Compute correlation with log target
corr = X_train.corrwith(y_log).abs().sort_values(ascending=False)
top20 = corr.head(20)
print(top20)

fig, ax = plt.subplots(figsize=(9, 7))
top20.plot.barh(ax=ax, color='steelblue')
ax.set_title('Top 20 |Correlation| with log(SalePrice)')
ax.invert_yaxis()
plt.tight_layout()
plt.savefig(FIG / '04_top_corr.png', bbox_inches='tight')
plt.show()
```

<div class="nb-section">
  <h2>5. OverallQual & TotalSF Interaction</h2>
</div>


```python
fig, ax = plt.subplots(figsize=(10, 6))
sc = ax.scatter(X_train['TotalSF'], y_log, c=X_train['OverallQual'], cmap='viridis', alpha=0.5, s=20)
plt.colorbar(sc, label='OverallQual')
ax.set_xlabel('TotalSF (log-scaled if applied)')
ax.set_ylabel('log(SalePrice)')
ax.set_title('TotalSF vs log(SalePrice) colored by OverallQual')
plt.tight_layout()
plt.savefig(FIG / '05_interaction.png', bbox_inches='tight')
plt.show()
```

<div class="nb-section">
  <h2>6. Feature Importance Preview (Random Forest)</h2>
</div>


```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score

# Quick RF for importance ranking (on a subset of features for speed)
rf = RandomForestRegressor(n_estimators=200, max_depth=12, random_state=42, n_jobs=-1)
rf.fit(X_train, y_log)

imp = pd.Series(rf.feature_importances_, index=X_train.columns).sort_values(ascending=False)
print(imp.head(25))

fig, ax = plt.subplots(figsize=(9, 8))
imp.head(25).plot.barh(ax=ax, color='teal')
ax.set_title('Random Forest Feature Importance (top 25)')
ax.invert_yaxis()
plt.tight_layout()
plt.savefig(FIG / '06_rf_importance.png', bbox_inches='tight')
plt.show()

# Quick CV score
scores = cross_val_score(rf, X_train, y_log, cv=5, scoring='neg_root_mean_squared_error', n_jobs=-1)
print(f'RF 5-fold CV RMSE (log): {-scores.mean():.4f} ± {scores.std():.4f}')
```

<div class="nb-section">
  <h2>7. Dashboard-style Summary Metrics</h2>
</div>


```python
summary = {
    'n_train': len(X_train),
    'n_features': X_train.shape[1],
    'target_mean': y.mean(),
    'target_median': y.median(),
    'target_std': y.std(),
    'log_skew': y_log.skew(),
    'top_feature': imp.index[0],
    'top_feature_imp': imp.iloc[0],
}
print(pd.Series(summary))

# Save importance for later
imp.to_csv('../Reports/rf_feature_importance.csv')
print('Saved rf_feature_importance.csv')
```
