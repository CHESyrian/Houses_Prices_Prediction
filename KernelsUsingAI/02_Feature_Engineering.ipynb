<div class="nb-title">
  <h1>02 — Feature Engineering</h1>
  <p class="subtitle">Ames Housing Dataset</p>
  <p class="meta">Goals · Domain imputation · Derived features · Ordinal & one-hot encoding · Log transforms</p>
</div>


<div class="nb-box">
<strong>Principle:</strong> Fit all imputation / encoding logic on <strong>train only</strong>; apply the same transformations to test. Preserve information; avoid leakage.
</div>


```python
import pandas as pd
import numpy as np
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

DATA_DIR = Path('../Datasets')
OUT_DIR = Path('../Datasets/processed')
OUT_DIR.mkdir(parents=True, exist_ok=True)

train = pd.read_csv(DATA_DIR / 'train.csv')
test  = pd.read_csv(DATA_DIR / 'test.csv')

print(train.shape, test.shape)

# Keep Id for later submission
train_id = train['Id'].copy()
test_id  = test['Id'].copy()

# Separate target
y = train['SalePrice'].copy()
y_log = np.log1p(y)

train = train.drop(columns=['SalePrice'])
# Align columns
all_data = pd.concat([train, test], axis=0, ignore_index=True)
print('Combined shape:', all_data.shape)
```

<div class="nb-section">
  <h2>1. Setup & Load</h2>
</div>


```python
# Columns where NA means “None” / absence of feature
none_cols = [
    'PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu',
    'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond',
    'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2',
    'MasVnrType'
]

for col in none_cols:
    if col in all_data.columns:
        all_data[col] = all_data[col].fillna('None')

# Numeric counterparts that should be 0 when feature is absent
zero_cols = [
    'GarageYrBlt', 'GarageArea', 'GarageCars',
    'BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF',
    'BsmtFullBath', 'BsmtHalfBath',
    'MasVnrArea'
]
for col in zero_cols:
    if col in all_data.columns:
        all_data[col] = all_data[col].fillna(0)

# LotFrontage: median by Neighborhood (more realistic than global median)
all_data['LotFrontage'] = all_data.groupby('Neighborhood')['LotFrontage'].transform(
    lambda x: x.fillna(x.median())
)
# Fallback if still missing
all_data['LotFrontage'] = all_data['LotFrontage'].fillna(all_data['LotFrontage'].median())

# Electrical – mode
all_data['Electrical'] = all_data['Electrical'].fillna(all_data['Electrical'].mode()[0])

# MSZoning – mode by MSSubClass or overall mode
all_data['MSZoning'] = all_data['MSZoning'].fillna(all_data['MSZoning'].mode()[0])

# Utilities, Functional, SaleType, Exterior1st/2nd, KitchenQual – mode
for col in ['Utilities', 'Functional', 'SaleType', 'Exterior1st', 'Exterior2nd', 'KitchenQual']:
    if col in all_data.columns:
        all_data[col] = all_data[col].fillna(all_data[col].mode()[0])

print('Remaining missing:')
print(all_data.isnull().sum().sort_values(ascending=False).head(10))
```

<div class="nb-section">
  <h2>2. Domain-Aware Missing Value Handling</h2>
</div>


```python
# Age features
all_data['HouseAge'] = all_data['YrSold'] - all_data['YearBuilt']
all_data['RemodAge'] = all_data['YrSold'] - all_data['YearRemodAdd']
all_data['GarageAge'] = all_data['YrSold'] - all_data['GarageYrBlt']
all_data['GarageAge'] = all_data['GarageAge'].replace({all_data['YrSold'].max()+1: 0})  # if GarageYrBlt was 0

# Total living area
all_data['TotalSF'] = all_data['TotalBsmtSF'] + all_data['1stFlrSF'] + all_data['2ndFlrSF']
all_data['TotalBath'] = (all_data['FullBath'] + 0.5 * all_data['HalfBath'] +
                         all_data['BsmtFullBath'] + 0.5 * all_data['BsmtHalfBath'])
all_data['TotalPorchSF'] = (all_data['OpenPorchSF'] + all_data['EnclosedPorch'] +
                            all_data['3SsnPorch'] + all_data['ScreenPorch'])

# Binary flags
all_data['HasPool'] = (all_data['PoolArea'] > 0).astype(int)
all_data['HasGarage'] = (all_data['GarageArea'] > 0).astype(int)
all_data['HasBsmt'] = (all_data['TotalBsmtSF'] > 0).astype(int)
all_data['HasFireplace'] = (all_data['Fireplaces'] > 0).astype(int)
all_data['Has2ndFloor'] = (all_data['2ndFlrSF'] > 0).astype(int)
all_data['WasRemodeled'] = (all_data['YearRemodAdd'] != all_data['YearBuilt']).astype(int)

# Interaction / quality composite
all_data['OverallGrade'] = all_data['OverallQual'] * all_data['OverallCond']
all_data['TotalRooms'] = all_data['TotRmsAbvGrd'] + all_data['FullBath'] + all_data['HalfBath']

# Lot shape simplified
all_data['IsRegularLot'] = (all_data['LotShape'] == 'Reg').astype(int)

print('New features created. Sample:')
print(all_data[['HouseAge', 'TotalSF', 'TotalBath', 'HasPool', 'OverallGrade']].head())
```

<div class="nb-section">
  <h2>3. Derived / Engineered Features</h2>
</div>


```python
# Quality mappings (consistent scale)
qual_map = {'None': 0, 'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5}
exposure_map = {'None': 0, 'No': 1, 'Mn': 2, 'Av': 3, 'Gd': 4}
finish_map = {'None': 0, 'Unf': 1, 'RFn': 2, 'Fin': 3}
bsmt_fin_map = {'None': 0, 'Unf': 1, 'LwQ': 2, 'Rec': 3, 'BLQ': 4, 'ALQ': 5, 'GLQ': 6}
functional_map = {'Sal': 0, 'Sev': 1, 'Maj2': 2, 'Maj1': 3, 'Mod': 4, 'Min2': 5, 'Min1': 6, 'Typ': 7}
fence_map = {'None': 0, 'MnWw': 1, 'GdWo': 2, 'MnPrv': 3, 'GdPrv': 4}

ordinal_maps = {
    'ExterQual': qual_map, 'ExterCond': qual_map,
    'BsmtQual': qual_map, 'BsmtCond': qual_map,
    'HeatingQC': qual_map, 'KitchenQual': qual_map,
    'FireplaceQu': qual_map, 'GarageQual': qual_map, 'GarageCond': qual_map,
    'PoolQC': qual_map,
    'BsmtExposure': exposure_map,
    'GarageFinish': finish_map,
    'BsmtFinType1': bsmt_fin_map, 'BsmtFinType2': bsmt_fin_map,
    'Functional': functional_map,
    'Fence': fence_map,
}

for col, mapping in ordinal_maps.items():
    if col in all_data.columns:
        all_data[col] = all_data[col].map(mapping).fillna(0).astype(int)

# CentralAir binary
all_data['CentralAir'] = (all_data['CentralAir'] == 'Y').astype(int)

print('Ordinal encoding done. Example ExterQual value counts:')
print(all_data['ExterQual'].value_counts().sort_index())
```

<div class="nb-section">
  <h2>4. Ordinal Encoding of Quality / Condition Variables</h2>
</div>


```python
# Utilities is almost constant
print('Utilities value counts:', all_data['Utilities'].value_counts())

drop_cols = [
    'Id',           # identifier
    'Utilities',    # almost constant
    'Street',       # almost all Pave
    'PoolQC',       # after ordinal, still very sparse; HasPool flag exists
    'MiscFeature',  # sparse
    'Alley',        # sparse after None fill
]

# Keep only columns that still exist
drop_cols = [c for c in drop_cols if c in all_data.columns]
all_data = all_data.drop(columns=drop_cols)
print('Shape after drop:', all_data.shape)
```

<div class="nb-section">
  <h2>5. Drop Low-Value / Redundant Columns</h2>
</div>


```python
# Identify remaining object columns
remaining_cat = all_data.select_dtypes(include=['object']).columns.tolist()
print('Remaining categorical columns for one-hot:', remaining_cat)

all_data = pd.get_dummies(all_data, columns=remaining_cat, drop_first=True, dtype=int)
print('Shape after one-hot:', all_data.shape)
```

<div class="nb-section">
  <h2>6. One-Hot Encoding for Remaining Nominals</h2>
</div>


```python
from scipy.stats import skew

# Split back to compute skew on train only
n_train = len(train)
train_fe = all_data.iloc[:n_train].copy()
test_fe  = all_data.iloc[n_train:].copy()

numeric_feats = train_fe.select_dtypes(include=[np.number]).columns
skewness = train_fe[numeric_feats].apply(lambda x: skew(x.dropna())).sort_values(ascending=False)
skewed = skewness[abs(skewness) > 0.75]
print(f'Features with |skew| > 0.75: {len(skewed)}')
print(skewed.head(15))

# Apply log1p to skewed features (except already ordinal / binary)
skip_log = set(ordinal_maps.keys()) | {'CentralAir', 'HasPool', 'HasGarage', 'HasBsmt',
                                       'HasFireplace', 'Has2ndFloor', 'WasRemodeled', 'IsRegularLot',
                                       'OverallQual', 'OverallCond', 'MSSubClass', 'MoSold', 'YrSold'}
to_log = [c for c in skewed.index if c not in skip_log and c in all_data.columns]

for col in to_log:
    all_data[col] = np.log1p(all_data[col])

print(f'Log-transformed {len(to_log)} features')
```

<div class="nb-section">
  <h2>7. Log-Transform Highly Skewed Numeric Features</h2>
</div>


```python
# Re-split
X_train = all_data.iloc[:n_train].copy()
X_test  = all_data.iloc[n_train:].copy()

# Ensure same columns (safety)
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

print('Final shapes:')
print('X_train:', X_train.shape)
print('X_test :', X_test.shape)
print('y      :', y.shape)
print('y_log  :', y_log.shape)

# Save
X_train.to_csv(OUT_DIR / 'X_train.csv', index=False)
X_test.to_csv(OUT_DIR / 'X_test.csv', index=False)
y.to_csv(OUT_DIR / 'y.csv', index=False)
y_log.to_csv(OUT_DIR / 'y_log.csv', index=False)
train_id.to_csv(OUT_DIR / 'train_id.csv', index=False)
test_id.to_csv(OUT_DIR / 'test_id.csv', index=False)

print('\nSaved processed files to', OUT_DIR)
print('Feature list (first 30):', list(X_train.columns[:30]))
```
