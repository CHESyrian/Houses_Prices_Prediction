<p style="height:60px;line-height:60px;background-color:navy;padding-left:12px;font-size:32px;font-family:serif;font-weight:bold;color:white;border:1px outset gold;border-radius:10px;">
    Nominal Variables
</p>

```python
houses[nom_features].nunique().sort_values()
```

```python
variables   = []
f_stats     = []
p_values    = []
etas_values = []
has_rel     = []
rel_level   = []

for col in nom_features:
    groups = [
        houses[houses[col] == level]['SalePrice'] 
        for level in houses[col].unique()
    ]
    
    f_stat, p = ss.f_oneway(*groups)
    
    grand_mean = houses['SalePrice'].mean()
    ss_between = sum([
        len(group) * (group.mean() - grand_mean)**2
        for group in groups
    ])
    
    ss_total = sum((houses['SalePrice'] - grand_mean)**2)
    eta_squared = ss_between / ss_total

    variables.append(col)
    f_stats.append(f_stat)
    p_values.append(p)
    etas_values.append(eta_squared)
    
    if p < 0.05:
        has_rel.append('Yes')
        if eta_squared >= 0.14:
            rel_level.append('Strong')
        elif 0.1 < eta_squared < 0.14:
            rel_level.append('Medium')
        else:
            rel_level.append('Weak')

    else:
        has_rel.append('No')
        rel_level.append('No')

nom_corr = pd.DataFrame(
    data={
        'Features'     : variables,
        'F Stats'      : f_stats,
        'P Values'     : p_values,
        'EtaSq Values' : etas_values,
        'Has Relation' : has_rel,
        'Effect Level' : rel_level
    }
)

nom_corr
```

**`MasVnrArea`, `BsmtFinSF1`, `2ndFlrSF`, `WoodDeckSF`, `OpenPorchSF`**

---

```python
spc_vars_1 = [
    'MasVnrArea', 
    'BsmtFinSF1', 
    '2ndFlrSF', 
    'WoodDeckSF', 
    'OpenPorchSF'
]
```

```python
variables = []
Pop_Means = []
S1_Means  = []
S2_Means  = []
B_Corr    = []
A_Corr    = []
t1_res    = []
t2_res    = []

for col in spc_vars_1:
        
    vrb_gt0 = houses[houses[col] > 0]
    vrb_eq0 = houses[houses[col] == 0]
    
    mu = houses['SalePrice'].mean()
    x_1 = vrb_gt0['SalePrice'].mean()
    x_2 = vrb_eq0['SalePrice'].mean()
    
    variables.append(col)
    Pop_Means.append(mu)
    S1_Means.append(x_1)
    S2_Means.append(x_2)
    B_Corr.append(h_corr.at[col, 'SalePrice'])
    A_Corr.append(vrb_gt0.corr(numeric_only=True).at[col, 'SalePrice'])
    
    stats_1, p_value_1 = ss.ttest_1samp(vrb_gt0['2ndFlrSF'], mu)
    stats_2, p_value_2 = ss.ttest_ind(vrb_gt0['2ndFlrSF'], vrb_eq0['2ndFlrSF'])
    
    if p_value_1 > 0.05:
        t1_res.append('Fail Reject H0')
        
    else:
        t1_res.append('Reject H0')
        
    if p_value_2 > 0.05:
        t2_res.append('Fail Reject H0')
        
    else:
        t2_res.append('Reject H0')

h_tests = pd.DataFrame({
    'Features' : variables, 
    'Population Mean' : Pop_Means,
    'Sample_1 Mean' : S1_Means,
    'Sample_2 Mean' : S2_Means,
    'Corr Befor del Zeros' : B_Corr,
    'Corr After del Zeros' : A_Corr,
    'TTest OneSample' : t1_res,
    'TTest IndSample' : t2_res
})
h_tests
```
